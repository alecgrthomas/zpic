In the lecture, we discussed the two stream instability, beam-plasma instability and Landau damping. The two-stream instability arises when two populations of charged particles move through each other at different velocities. Slight disturbances in the electric field cause the particles to bunch together, increasingly amplifying these disturbances and ultimately resulting in growing electrostatic waves. In astrophysics, the two-stream instability is seen in contexts such as collisionless shocks (like those in supernova remnants), the solar wind, and the streaming of cosmic rays through interstellar plasma. In laboratory settings, it can occur when particle beams are injected into plasmas, as in certain fusion devices. The idealized version of these is two beams of equal density counterpropagating at fluid momenta $\pm p_0$. A related phenomenon is the beam-plasma instability, in which a dilute, high-momentum cold beam (low momentum spread) of charged particles, such as electrons or ions, penetrates a background plasma, which may be hot (have a large momentum spread). Beam-plasma instabilities play a key role in astrophysical processes like auroral particle precipitation and in the interaction of relativistic jets with the interstellar medium. They are also observed in laboratory plasmas when external particle beams are used for heating or diagnostic purposes, such as neutral beam injection in a tokamak. Landau damping seems superficially very different; a mechanism by which the amplitude of plasma waves decreases without collisions. This occurs when particles with velocities close to the wave’s phase velocity exchange energy with the wave, gradually reducing its amplitude. Landau damping is crucial for regulating turbulence and energy dissipation in cosmic plasmas, such as the solar wind, and also plays an important role in the stability and heating of laboratory plasmas, including in fusion experiments. 

What links these phenomena is that they involve the coupling of electrostatic waves generated by the bulk plasma motion to particle acceleration (or deceleration), which can only be understood in the context of kinetics. All involve a phase-space vortex for particles that involves particles being trapped in the potential of the plasma. The two stream instability is by far the easiest to initiate, and can be done with a “cold” distributions, i.e. without the need for a thermal spread. ### Landau damping of a thermal traveling wave

In the lectures we showed that the linear growth rate of the Landau instability for a warm plasma is 
$$
\gamma \simeq \sqrt{\frac{\pi}{8}}\frac{1}{(k\lambda_D)^3}\exp\left(-\frac{1}{k^2\lambda_D^2}-3\right)
$$
with $\lambda_D = v_{th}/\omega_p = v_{th}$ in our normalization. **Note that this damping rate is defined for the electric field energy $E^2/2$, not the density perturbation as we had before.** 

We are going to study this important plasma concept with ```zpic```. You should decide yourself on the correct parameters. **Note that this is a very tricky problem for the particle-in-cell code!** For Landau damping you will need to initiate a \emph{warm traveling wave}, as in Labscript 5 (start from there). However, you need to ensure that Landau damping can occur, which requires the velocity gradient of the particle distribution function, $\partial f/\partial v$ to be significant near the phase velocity of the wave $\omega/k$. You should use the analytic theory expression given above to guide you. Note that: 

* The linear analytic theory given above is derived under the assumption that $k\lambda_D<1$. For the enthusiastic, you could numerically solve the dispersion function given in the lectures to examine the arbitrary case, however....
* also note that $v_{th}$, the width of the distribution, can't be too wide compared to the perturbation amplitude or else noise (thermal fluctuations) will be too large to distinguish the damping effect. 
* The above conditions  motivates studying large values of $k$ (i.e. slow phase velocity waves). Pushing the linear approximation a little, such that the density perturbation is large, $\delta n/n_0\sim0.05$, is recommended, to maximize the wave amplitude above noise (but still having linear behavior).  
* You will probably need many 1000s of particles-per-cell to get a reasonable result! Start with low resolution to initialize the warm plasma wave and then increase resolution so you have better particle statistics near the phase velocity of the wave, $v_p=\omega/k$.
